Using the list of S&P 500 companies from Wikipedia's S&P 500 companies page (https://en.wikipedia.org/wiki/List_of_S%26P_500_companies), download the data including the year each company was added to the index.

Steps:

1. Create a DataFrame with company tickers, names, and the year they were added.
2. Extract the year from the addition date and calculate the number of stocks added each year.
3. Which (full) year had the highest number of additions, starting from 2020?

▎ Note: When stocks are added to the S&P 500, they usually experience a price bump as investors and index funds buy shares following the announcement.

▎ Context:

"Following the announcement, all four new entrants saw their stock prices rise in extended trading on Friday" - recent examples of S&P 500 additions include DASH, WSM, EXE, TKO in 2025 (Nasdaq article - https://www.nasdaq.com/articles/sp-500-reshuffle-dash-tko-expe-wsm-join-worth-buying).

*Additional*: How many current S&P 500 stocks have been in the index for more than 20 years?

````python
# IMPORTS
import pandas as pd
import requests
from datetime import datetime
````

````python
url = 'https://en.wikipedia.org/wiki/List_of_S%26P_500_companies'
# 1. Define custom headers to mimic a real web browser
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'}

# Fetch the HTML content with headers
response = requests.get(url, headers=headers)

# 3. Pass the HTML text string to pandas
# read_html always returns a LIST of DataFrames (one for each <table> found)
tables = pd.read_html(response.text)

# 4. Extract the specific table you want (e.g., the first table)
sp500 = tables[0]
sp500 = sp500[["Symbol", "Security", "Date added"]]

# 1. Convert column to datetime
sp500['Date added'] = pd.to_datetime(sp500['Date added'], errors="coerce")

# 2. Extract the year
sp500['Year added'] = sp500['Date added'].dt.year

sp500.head()
````

````python
stocks_since_2020 = sp500[sp500["Year added"] >= 2020]

stocks_since_2020["Year added"].value_counts().sort_index()
````

````python
stocks_since_2020["Year added"].value_counts().idxmax()
````

**Answer**:
2025

````python
# How many current S&P 500 stocks have been in the index for more than 20 years?
today = pd.Timestamp.today().normalize()
cutoff = today - pd.DateOffset(years=20)

more_than_20_years = sp500[sp500["Date added"] < cutoff]

len(more_than_20_years)
````

**Answer**: 224

**Question 3. [Index] S&P 500 Market Corrections Analysis**

**Calculate the median drawdown (in %) of significant market corrections in the S&P 500 index.**

For this task, define a correction as an event when a stock index goes down by **at least 5%** from the most recent all-time high.

Steps:

1. Download S&P 500 historical data (period from 1950 to present) using yfinance (daily stats)
2. Identify all-time high closing price points (where price exceeds all previous days' prices)
3. For each pair of consecutive all-time highs, find the minimum price in between
4. Calculate drawdown percentages: (high - low) / high × 100
5. Filter for corrections with at least 5% drawdown
6. Calculate the duration in days for each correction period
7. Determine the 25th, 50th (median), and 75th percentiles for correction durations and drawdowns

Context:

- Investors often wonder about the typical length of market corrections when deciding "when to buy the dip" (Reddit discussion - https://www.reddit.com/r/investing/comments/1jrqnte/when_are_you_buying_the_dip/?rdt=64135).
- A Wealth of Common Sense - How Often Should You Expect a Stock Market Correction? - https://awealthofcommonsense.com/2022/01/how-often-should-you-expect-a-stock-market-correction/


*Hint (use this data to compare with your results)*: Here is the list of top 10 largest corrections by drawdown:

- 2007-10-09 to 2009-03-09: 56.8% drawdown over 517 days
- 2000-03-24 to 2002-10-09: 49.1% drawdown over 929 days
- 1973-01-11 to 1974-10-03: 48.2% drawdown over 630 days
- 1968-11-29 to 1970-05-26: 36.1% drawdown over 543 days
- 2020-02-19 to 2020-03-23: 33.9% drawdown over 33 days
- 1987-08-25 to 1987-12-04: 33.5% drawdown over 101 days
- 1961-12-12 to 1962-06-26: 28.0% drawdown over 196 days
- 1980-11-28 to 1982-08-12: 27.1% drawdown over 622 days
- 2022-01-03 to 2022-10-12: 25.4% drawdown over 282 days
- 1966-02-09 to 1966-10-07: 22.2% drawdown over 240 days

````python
# Download S&P 500 historical data (period from 1950 to present) using yfinance (daily stats)
start_date = "1950-01-01"

ticker_obj = yf.Ticker("^SPX")

sp500 = ticker_obj.history(
    start=start_date,
    interval="1d",
    auto_adjust=False
)

# make the entire DataFrame timezone-naive
sp500.index = sp500.index.tz_localize(None)

sp500 = sp500[["Close"]].copy()

sp500.head(15)
````

````python
# Identify all-time high closing price points (where price exceeds all previous days' prices)

# calculate the highest closing price seen up to each day
sp500["Running High"] = sp500["Close"].cummax()
sp500.head(15)

````

````python
# explicitly identify when the price exceeds all previous closes
sp500["Is ATH"] = sp500["Close"] > sp500["Running High"].shift(1)

sp500.loc[sp500.index[0], "Is ATH"] = True

ath_dates = sp500.index[sp500["Is ATH"]].tolist()

sp500.head(15)
````

````python
# For each pair of consecutive all-time highs, find the minimum price in between
corrections = []

for i in range(len(ath_dates) - 1):
  high_date = ath_dates[i]
  next_high_date = ath_dates[i + 1]

  period = sp500.loc[high_date:next_high_date]

  high_price = sp500.loc[high_date, "Close"]

  low_price = period["Close"].min()
  low_date = period["Close"].idxmin()

  # Calculate drawdown percentages: (high - low) / high × 100
  drawdown = (
    (high_price - low_price)
    / high_price
    * 100
  )

  # Filter for corrections with at least 5% drawdown
  if drawdown >= 5:

    # Calculate the duration in days for each correction period
    duration = (low_date - high_date).days

    corrections.append({
      "High Date": high_date,
      "Low Date": low_date,
      "High": high_price,
      "Low": low_price,
      "Drawdown %": drawdown,
      "Duration Days": duration
    })

corrections_df = pd.DataFrame(corrections)
corrections_df.sort_values("Drawdown %", ascending=False).head(10)
````

````python
# Determine the 25th, 50th (median), and 75th percentiles for correction durations and drawdowns
percentiles = corrections_df[
    ["Drawdown %", "Duration Days"]
].quantile([0.25, 0.50, 0.75])

percentiles
````

````python
corrections_df["Drawdown %"].median()
````

**Question 4. [Stocks] Earnings Surprise Analysis for Amazon (AMZN)**

**Calculate the median 2-day percentage change in stock prices following positive earnings surprise days.**

Steps:

1. Load earnings data for the last few years using the yfinance method get_earnings_dates():

```
import yfinance as yf

ticker = 'AMZN'
ticker_obj = yf.Ticker(ticker)
result = ticker_obj.get_earnings_dates()
```

You should have 25 entries starting from 2020-10-29.

Note: the 1 future earnings date entry will not have Reported EPS
and Surprise % filled in.

2. Download complete historical price data using yfinance
3. Calculate 2-day percentage changes for all historical dates: for each sequence of 3 consecutive trading days (Day 1, Day 2, Day 3), compute the *return* as Close_Day3 / Close_Day1 - 1. (Assume Day 2 may correspond to the earnings announcement.)
4. Filter for positive earnings surprises and calculate the median 2-day return. Then calculate the correlation between the 2-day stock return and the earnings surprise magnitude.

▎ *Hint*: you can generate all correlations at once using pd.corr().

**Context**: Earnings announcements, especially when they exceed analyst expectations, can significantly impact stock prices in the short term.

**Reference**: Yahoo Finance Earnings Calendar -  https://finance.yahoo.com/calendar/earnings?symbol=AMZN

*Additional*: Is there a correlation between the magnitude of the earnings surprise and the stock price reaction? Does the market react differently to earnings surprises during bull vs. bear markets?

````python
# Load earnings data for the last few years using the yfinance method get_earnings_dates()
ticker = 'AMZN'
ticker_obj = yf.Ticker(ticker)
earnings = ticker_obj.get_earnings_dates()

earnings = earnings.dropna(
  subset=["Reported EPS", "Surprise(%)"]
).copy()

# normalize dates
earnings["Date"] = earnings.index.date

earnings
````

````python
# Download complete historical price data using yfinance
start_date = "2020-01-01"

prices = ticker_obj.history(
    start=start_date,
    interval="1d",
    auto_adjust=False
)

prices = prices[["Close"]].copy()

# normalize dates
prices["Date"] = prices.index.date

prices.index.name = None

prices.head(15)
````

````python
# Calculate 2-day percentage changes for all historical dates:
# for each sequence of 3 consecutive trading days (Day 1, Day 2, Day 3),
# compute the return as Close_Day3 / Close_Day1 - 1.
# (Assume Day 2 may correspond to the earnings announcement.)

prices["2d_return"] = prices["Close"].shift(-1) / prices["Close"].shift(1) - 1

prices
````

````python
# Match earnings dates to price dates
earnings_with_returns = earnings.merge(
  prices[["Date", "2d_return"]],
  on="Date",
  how="left"
)

earnings_with_returns
````

````python
# Filter for positive earnings surprises
positive = earnings_with_returns[
  earnings_with_returns["Surprise(%)"] > 0
].copy()
````

````python
positive
````

````python
# Calculate the median 2-day return
median_return = positive["2d_return"].median() * 100
median_return
````

````python
correlation = earnings_with_returns[
    ["Surprise(%)", "2d_return"]
].corr().iloc[0, 1]

correlation
````

````python
print(f"Median 2-day return: {median_return:.2f}%")
print(f"Correlation: {correlation:.3f}")
````

There is a weak positive correlation of about 0.22, suggesting that larger positive earnings surprises are associated with slightly stronger stock-price reactions, but the relationship is not strong.

The median 2-day return following positive earnings surprises is about +0.35%, indicating that the typical short-term price reaction is positive but modest.
